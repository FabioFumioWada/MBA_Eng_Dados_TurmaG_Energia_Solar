```python
import requests

# ============================================================
# CONFIGURAÇÕES
# ============================================================
url = "https://dadosabertos.aneel.gov.br/dataset/6d90b77c-c5f5-4d81-bdec-7bc619494bb9/resource/11ec447d-698d-4ab8-977f-b424d5deee6a/download/siga-empreendimentos-geracao.csv"

arquivo_volume = "/Volumes/mba/stage/dados_bruto/usinas/siga-empreendimentos-geracao.csv"


# ============================================================
# DOWNLOAD DO ARQUIVO
# ============================================================
response = requests.get(url, timeout=120)
response.raise_for_status()


# ============================================================
# SALVA O ARQUIVO ORIGINAL NO VOLUME
# ============================================================
with open(arquivo_volume, "wb") as arquivo:
    arquivo.write(response.content)


print("Arquivo baixado com sucesso!")
print(f"Tamanho: {len(response.content):,} bytes")
```
