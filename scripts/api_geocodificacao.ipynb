```python
import requests
import json

# ============================================================
# API NOMINATIM - REVERSE GEOCODING
# ============================================================

def obter_localizacao(latitude, longitude):
    url = "https://nominatim.openstreetmap.org/reverse"

    params = {
        "lat": latitude,
        "lon": longitude,
        "format": "jsonv2",
        "addressdetails": 1,
        "accept-language": "pt-BR"
    }

    headers = {
        "User-Agent": "Databricks-Clima-INMET/1.0"
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=30
    )

    response.raise_for_status()

    dados = response.json()
    endereco = dados.get("address", {})
    #bairro = endereco.get("suburb")
    #cidade = endereco.get("city")
    estado = endereco.get("state")
    regiao = endereco.get("region")
    cidade = endereco.get('city') or endereco.get('town') or endereco.get('suburb') or "Não encontrada"
    bairro = endereco.get('suburb') or endereco.get('neighbourhood') or endereco.get('road') or "Zona Rural / Não mapeado"

    return {
                "bairro": bairro,
                "cidade": cidade,
                "estado": estado,
                "regiao": regiao
            }

#teste  
#latitude = -23.5358
#longitude = -46.5097
#resultado = obter_localizacao(
#    latitude,
#    longitude
#)
#print(resultado)
```
