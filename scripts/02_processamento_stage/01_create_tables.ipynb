```python
%sql
-- ============================================
-- RECRIAÇÃO DA TABELA DELTA
-- Tabela: conta_bandeira
-- Camada: RAW
-- Fonte: ANEEL
-- ============================================

CREATE TABLE IF NOT EXISTS mba.raw.conta_bandeira
(
    DatGeracaoConjuntoDados DATE
        COMMENT 'Data do processamento de carga automática no momento da geração para publicação do conjunto de dados abertos.',

    SigAgente STRING
        COMMENT 'Sigla que abrevia o nome dos Agentes regulados pela ANEEL.',

    NumCPFCNPJ STRING
        COMMENT 'CNPJ do Agente do setor elétrico conforme cadastro de agentes da ANEEL.',

    DatCompetencia DATE
        COMMENT 'Data de competência da apuração dos repasses da Conta Bandeira.',

    VlrReceitaFaturada DECIMAL(21,5)
        COMMENT 'Valor faturado pelas distribuidoras com a aplicação dos adicionais da conta bandeira nas tarifas dos consumidores. Valores informados pelas distribuidoras por meio do SAMP.',

    VlrRepasseContaBandeira DECIMAL(21,5)
        COMMENT 'Valores dos repasses da Conta Bandeira. Valores positivos são recebimentos de recursos pela distribuidora e valores negativos são repasses de recursos pela distribuidora para a conta. Valores definidos pela ANEEL após a apuração dos custos de geração de energia pela CCEE.',

    VlrResultadoMCP DECIMAL(21,5)
        COMMENT 'Custos com Mercado de Curto Prazo - MCP para o mês de competência, apurados e informados pela CCEE.',

    VlrCCEARD DECIMAL(21,5)
        COMMENT 'Custos com efeito de disponibilidade dos contratos de CCEAR-D para o mês de competência, apurados e informados pela CCEE.',

    VlrRiscoHidroCCGFRepactuadas DECIMAL(21,5)
        COMMENT 'Custos com Risco Hidrológico associado aos contratos CCGF e das usinas repactuadas para o mês de competência, apurados e informados pela CCEE. Esse valor é obtido após a dedução do prêmio de risco das usinas repactuadas e da previsão de risco hidrológico concedida no processo tarifário.',

    VlrRiscoHidrologicoItaipu DECIMAL(21,5)
        COMMENT 'Custos de Risco Hidrológico associado à Itaipu para o mês de competência, apurados e informados pela CCEE.',

    VlrRiscoHidrologicoRepactuadas DECIMAL(21,5)
        COMMENT 'Custos com Risco Hidrológico associado às usinas repactuadas para o mês de competência, apurados e informados pela CCEE.',

    VlrRiscoHidrologicoCCGF DECIMAL(21,5)
        COMMENT 'Custos com Risco Hidrológico associado aos contratos CCGF para o mês de competência, apurados e informados pela CCEE.',

    VlrPrevisaoRiscoHidrologico DECIMAL(21,5)
        COMMENT 'Previsão de Risco Hidrológico considerado nos processos tarifários das distribuidoras.',

    VlrPremiodeRisco DECIMAL(21,5)
        COMMENT 'Valor do prêmio de risco associado aos contratos das usinas repactuadas.',

    VlrESSEER DECIMAL(21,5)
        COMMENT 'Custos com Encargos de Serviço de Sistemas - ESS e Encargos de Energia de Reserva - EER para o mês de competência, apurados e informados pela CCEE.',

    VlrRessarcimentoCONER DECIMAL(21,5)
        COMMENT 'Recebimento do ressarcimento da CONER para o mês de competência, apurado e informado pela CCEE.',

    VlrCVAEnergiaMesAnterior DECIMAL(21,5)
        COMMENT 'Custos acumulados não amortizados com CVA Energia para o mês anterior ao de competência da apuração da Conta Bandeira.',

    VlrESSEERMesAnterior DECIMAL(21,5)
        COMMENT 'Custos acumulados não amortizados com Encargos de Serviço de Sistemas - ESS e Encargos de Energia de Reserva - EER para o mês anterior ao de competência da apuração da Conta Bandeira.',

    VlrExposicaoInvoluntariaMesAnt DECIMAL(21,5)
        COMMENT 'Custos acumulados não amortizados com exposição involuntária no mercado de curto prazo para o mês anterior ao de competência da apuração da Conta Bandeira.',

    VlrCVAEnergiaReceitaAlocada DECIMAL(21,5)
        COMMENT 'Receita alocada para a cobertura dos custos de CVA de Energia no mês de competência.',

    VlrCVAESSEERReceitaAlocada DECIMAL(21,5)
        COMMENT 'Receita alocada para a cobertura dos custos de ESS e EER no mês de competência.',

    VlrExposicaoInvolunReceitaAloc DECIMAL(21,5)
        COMMENT 'Receita alocada para a cobertura dos custos de exposição involuntária no mês de competência.',

    VlrCVAEnergiaAposRepasse DECIMAL(21,5)
        COMMENT 'Custos acumulados não amortizados com CVA Energia após repasse.',

    VlrCVAESSEERAposRepasse DECIMAL(21,5)
        COMMENT 'Custos acumulados não amortizados com Encargos de Serviço de Sistemas - ESS e Encargos de Energia de Reserva - EER após repasse.',

    VlrExposicaoInvolunAposRepasse DECIMAL(21,5)
        COMMENT 'Custos acumulados não amortizados com exposição involuntária no mercado de curto prazo após repasse.',

    VlrAjusteRepasse DECIMAL(21,5)
        COMMENT 'Ajuste para repasses menores do que R$ 1,00.'
)
USING DELTA
COMMENT 'Informações dos valores repassados pelos agentes de distribuição a partir do recolhimento dos recursos provenientes do faturamento das Bandeiras Tarifárias, em nome da Conta de Desenvolvimento Energético - CDE, diretamente à Conta Bandeira.';
```

```python
%sql
-- ============================================
-- CRIAÇÃO DAS TABELAS DELTA
-- Fonte: IBGE
-- Assunto: Estimativas da População - Brasil e Unidades da Federação
-- Camada: RAW
-- ============================================

CREATE TABLE IF NOT EXISTS mba.raw.ibge_populacao_brasil_ufs
(
    AnoReferencia INT
        COMMENT 'Ano de referência da estimativa populacional, extraído do nome do arquivo IBGE.',

    BrasilUnidadeFederacao STRING
        COMMENT 'Nome do Brasil, região ou Unidade da Federação referente à estimativa populacional.',

    PopulacaoEstimada BIGINT
        COMMENT 'Quantidade estimada de habitantes para o Brasil, região ou Unidade da Federação.',

    NomeArquivo STRING
        COMMENT 'Nome do arquivo de origem utilizado na carga.',

    DataCarga TIMESTAMP
        COMMENT 'Data e horário em que o registro foi carregado na tabela Delta.'
)
USING DELTA
COMMENT 'Estimativas da população residente no Brasil e Unidades da Federação, conforme arquivos divulgados pelo IBGE.';
```
