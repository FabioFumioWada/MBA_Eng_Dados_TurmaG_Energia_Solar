```python
%sql
select  substring_index(trim(CodCEG), '-', 1) as CodCEG, NomEmpreendimento, DscOrigemCombustivel, DscFonteCombustivel, DscTipoOutorga, NomFonteCombustivel, NumCoordNEmpreendimento as Latitude, NumCoordEEmpreendimento as Longitude, DscMuninicpios, SigUFPrincipal, MdaPotenciaOutorgadaKw as Potencia
from mba.raw.usinas
```

```python
%sql
SELECT
    a.NomEmpreendimento,
    a.CodCEG,
    b.nom_usina,
    b.ceg
FROM mba.raw.usinas a
JOIN mba.raw.ons_geracao_usina_horaria b

    ON substring_index(trim(a.CodCEG), '-', 1)
     =
       substring_index(trim(b.ceg), '-', 1)
```
