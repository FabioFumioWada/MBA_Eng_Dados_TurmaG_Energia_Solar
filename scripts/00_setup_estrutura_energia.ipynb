```python
%sql

-- ============================================================
-- SCRIPT: 00_setup_estrutura_energia
-- OBJETIVO:
-- Criar a estrutura inicial do domínio de dados "energia".
--
-- Este notebook deve ser executado antes dos notebooks de
-- ingestão e transformação de dados.
--
-- ESTRUTURA CRIADA:
-- Catalog : mba
-- Schema  : energia
--
-- Volumes:
--   raw     -> Arquivos recebidos em seu formato original.
--   stage   -> Arquivos temporários ou intermediários.
--   refined -> Arquivos tratados e padronizados.
--   trusted -> Arquivos finais prontos para consumo.
-- ============================================================

-- ------------------------------------------------------------
-- 1. CRIAÇÃO DO CATALOG
-- ------------------------------------------------------------

-- O Catalog é o nível mais alto da hierarquia do Unity Catalog.
--
-- Hierarquia:
-- Catalog
--    └── Schema
--          └── Tabelas / Views / Volumes

CREATE CATALOG IF NOT EXISTS mba;

-- ------------------------------------------------------------
-- 1. CRIAÇÃO DO SCHEMA
-- ------------------------------------------------------------

-- O schema é utilizado para organizar os objetos relacionados
-- ao domínio de energia, como tabelas, views e volumes.

CREATE SCHEMA IF NOT EXISTS mba.energia;


-- ------------------------------------------------------------
-- 2. CRIAÇÃO DOS VOLUMES
-- ------------------------------------------------------------

-- VOLUME: RAW
-- Camada destinada ao armazenamento dos arquivos originais
-- recebidos das fontes de origem.
--
-- Boa prática:
-- Evitar alterar os arquivos originais armazenados nesta camada.

CREATE VOLUME IF NOT EXISTS mba.energia.raw;


-- VOLUME: STAGE
-- Camada destinada ao armazenamento temporário de arquivos
-- durante etapas intermediárias do processamento.
--
-- Pode ser utilizada para arquivos auxiliares, dados
-- temporários ou resultados intermediários de transformações.

CREATE VOLUME IF NOT EXISTS mba.energia.stage;


-- VOLUME: REFINED
-- Camada destinada ao armazenamento de arquivos que já
-- passaram por processos de tratamento e padronização.
--
-- Exemplos:
-- - Correção de tipos de dados
-- - Padronização de nomes
-- - Tratamento de valores nulos
-- - Remoção de dados inválidos

CREATE VOLUME IF NOT EXISTS mba.energia.refined;


-- VOLUME: TRUSTED
-- Camada destinada aos arquivos finais, validados e prontos
-- para serem utilizados por processos de consumo ou análises.
--
-- Os dados desta camada devem possuir regras de qualidade
-- aplicadas e estar preparados para consumo.

CREATE VOLUME IF NOT EXISTS mba.energia.trusted;


-- ============================================================
-- FIM DO SETUP
```

```python
%sql
DROP VOLUME IF EXISTS mba.energia.raw;

DROP VOLUME IF EXISTS mba.energia.stage;

DROP VOLUME IF EXISTS mba.energia.refined;

DROP VOLUME IF EXISTS mba.energia.trusted;
```
