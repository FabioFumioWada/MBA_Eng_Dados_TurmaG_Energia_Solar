
## 00 - Setup da Estrutura do Domínio Energia

### Objetivo

Este notebook é responsável pela **criação e configuração da estrutura inicial do domínio de dados `energia`** no Databricks.

Deve ser executado **antes dos notebooks de ingestão e transformação**, garantindo que toda a estrutura necessária esteja disponível.

### Estrutura criada

**Catalog:** `mba`  
**Schema:** `energia`

### Volumes

| Volume | Finalidade |
|---|---|
| `stage` | Arquivos recebidos em seu formato original. |
| `raw` | Arquivos temporários ou intermediários do processamento. |
| `refined` | Dados tratados e padronizados. |
| `trusted` | Dados finais, validados e prontos para consumo. |

### Fluxo de processamento

`Setup → Raw → Stage → Refined → Trusted`

```python
%sql
-- ------------------------------------------------------------
-- 1. CRIAÇÃO DO CATALOG
-- ------------------------------------------------------------

-- O Catalog é o nível mais alto da hierarquia do Unity Catalog.
--
-- Hierarquia:
-- Catalog
--    └── Schema
--          └── Tabelas / Views / Volumes

CREATE CATALOG IF NOT EXISTS mba;

-- ------------------------------------------------------------
-- 2. CRIAÇÃO DO SCHEMA
-- ------------------------------------------------------------

-- O schema é utilizado para organizar os objetos relacionados
-- ao domínio de energia, como tabelas, views e volumes.

-- SCHEMA: STAGE
-- Camada destinada ao armazenamento dos arquivos originais
-- recebidos das fontes de origem.
--
-- Boa prática:
-- Evitar alterar os arquivos originais armazenados nesta camada.

CREATE SCHEMA IF NOT EXISTS mba.raw;


-- SCHEMA: RAW
-- Pode ser utilizada para arquivos auxiliares, dados
-- temporários ou resultados intermediários de transformações.

CREATE SCHEMA IF NOT EXISTS mba.stage;

-- Criação do volume RAW
CREATE VOLUME IF NOT EXISTS mba.stage.dados_bruto;
-- Adicionado por Fábio em 29/08/2026 -- dados antigos
--CREATE VOLUME IF NOT EXISTS mba.stage.dados_bruto.old;
-- Adicionado por Fábio em 29/08/2026 -- dados ANEEL
--CREATE VOLUME IF NOT EXISTS mba.stage.dados_bruto.ANEEL;
-- Adicionado por Fábio em 29/08/2026 -- dados ONS
--CREATE VOLUME IF NOT EXISTS mba.stage.dados_bruto.ONS;
-- Adicionado por Fábio em 29/08/2026 -- dados IBGE
--CREATE VOLUME IF NOT EXISTS mba.stage.dados_bruto.IBGE;

-- SCHEMA: REFINED
-- Camada destinada ao armazenamento de arquivos que já
-- passaram por processos de tratamento e padronização.
--
-- Exemplos:
-- - Correção de tipos de dados
-- - Padronização de nomes
-- - Tratamento de valores nulos
-- - Remoção de dados inválidos

CREATE SCHEMA IF NOT EXISTS mba.refined;


-- SCHEMA: TRUSTED
-- Camada destinada aos arquivos finais, validados e prontos
-- para serem utilizados por processos de consumo ou análises.
--
-- Os dados desta camada devem possuir regras de qualidade
-- aplicadas e estar preparados para consumo.

CREATE SCHEMA IF NOT EXISTS mba.trusted;


-- ============================================================
-- FIM DO SETUP
```

```python
%sh

mkdir -p /Volumes/mba/stage/dados_bruto/ANEEL
mkdir -p /Volumes/mba/stage/dados_bruto/IBGE
mkdir -p /Volumes/mba/stage/dados_bruto/ONS
```
